# Inspect one raw H5 file keys

Open one H5 file directly with `h5py`. No `load_raw_h5`, no filtering.

```python
from pathlib import Path
import h5py
```

```python
repo_root = None
for p in [Path.cwd().resolve()] + list(Path.cwd().resolve().parents):
    if (p / "analysis" / "scripts").exists():
        repo_root = p
        break
if repo_root is None:
    raise RuntimeError("Could not locate repository root containing analysis/scripts")

# Change only these if needed.
RUN_NO = 58780
FILE_INDEX = 0

raw_dir = repo_root / "11022188" / "raw" / "hdf" / "online-0" / "fl2user1"
files = sorted(raw_dir.glob(f"*run{RUN_NO}*.h5"))

print("cwd      :", Path.cwd())
print("repo_root:", repo_root)
print("raw_dir  :", raw_dir)
print("n files  :", len(files))
for i, f in enumerate(files[:20]):
    print(f"{i:3d}: {f.name}")

if not files:
    raise FileNotFoundError(f"No raw H5 files found for run {RUN_NO} in {raw_dir}")

H5_PATH = files[FILE_INDEX]
print("\nUsing:", H5_PATH)
```

```python
# Top-level keys only.
with h5py.File(H5_PATH, "r") as f:
    for k in f.keys():
        print(k)
```

```python
# All keys recursively, with shape and dtype for datasets.
def show(name, obj):
    if isinstance(obj, h5py.Dataset):
        print(f"{name}  shape={obj.shape}  dtype={obj.dtype}")
    else:
        print(f"{name}/")

with h5py.File(H5_PATH, "r") as f:
    f.visititems(show)
```
