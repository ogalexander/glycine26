# Static-XAS ML Quicklook

Run `python ML/scripts/run_pipeline.py --name <run_name>` first, then set `RUN_NAME` below.

```python
from pathlib import Path
import pandas as pd
from IPython.display import Image, display

ROOT = Path.cwd()
if ROOT.name == 'notebooks':
    ROOT = ROOT.parent
elif (ROOT / 'ML').exists():
    ROOT = ROOT / 'ML'
print(ROOT)
```

```python
RUN_NAME = ''  # e.g. '20260716_120000'; blank uses newest result folder

results_root = ROOT / 'results'
if RUN_NAME:
    RESULTS_DIR = results_root / RUN_NAME
else:
    candidates = sorted([p for p in results_root.iterdir() if p.is_dir()], key=lambda p: p.stat().st_mtime)
    if not candidates:
        raise FileNotFoundError(f'No result folders in {results_root}')
    RESULTS_DIR = candidates[-1]
print(RESULTS_DIR)
```

```python
heldout = pd.read_csv(RESULTS_DIR / 'metrics' / 'heldout_reference_metrics.csv')
absorbance = pd.read_csv(RESULTS_DIR / 'metrics' / 'binned_absorbance_metrics.csv')
display(heldout)
display(absorbance.head())
```

```python
for name in [
    'pca_explained_variance.png',
    'heldout_reference_predicted_vs_true.png',
    'heldout_reference_error_histograms.png',
    'nominal_energy_binned_ml_absorbance.png',
    'actual_energy_binned_ml_absorbance.png',
]:
    path = RESULTS_DIR / 'plots' / name
    if path.exists():
        print(name)
        display(Image(filename=str(path)))
```
